```python
"""
Note: You may have to run all cells in order for the code to work.

I just want to appreciate you guys for the opportunity to learn.
Doing the projects have really made me learn a lot.
Thank you very much.
"""

wells = {}    # Wells and their data

# Created the variables below to make it easier to call data needed for the calculations
initial_production_rates = []     #Initial production rates of wells
well = []     # name of wells
years = []     # Time for each well
decline_rates = []    # Decline rate of each well
wells_production_rates = {}     # Wells and their calculated production rates
```

```python
def input_parameters():
    name_of_well = input('What is the name of the well?: ')
    initial_production_rate = float(input('What is the initial production rate?: '))
    decline_rate = float(input('What is the decline rate?: '))
    time = int(input('Time(In years)?: ')) #time is in years.

    # The codes below insert data in different dictionaries and lists.
    wells[name_of_well] = [initial_production_rate, decline_rate, time]
    well.append(name_of_well.capitalize())
    initial_production_rates.append(initial_production_rate)
    years.append(time)
    decline_rates.append(decline_rate)
```

```python
def simulate_well():
    import math
    def arps_model(initial_rate, decline_rate, time):  # Define ARPS model
        """
        Function to implement Arp's exponential decline model.

        Arguments:
        initial_rate (float): Initial production rate.
        decline_rate (float): Decline rate.
        time (integer): Time since the start of production.

        Returns:
        production_rate (float): Production rate at specified time.
        """
        return initial_rate * math.exp(-decline_rate * time)

    def simulate_production():
        for well_name, data in wells.items():
            qi = data[0]
            d = data[1]
            t = data[2]
            print('Well: ', well_name.capitalize())
            production_rates = []
            for year in range(t + 1):
                arps_model_calc = arps_model(qi, d, year)
                production_rates.append(round(arps_model_calc, 2))
                print("Year:", year, '-', "Production Rate:", round(arps_model_calc, 2))
            wells_production_rates[well_name.capitalize()] = production_rates
    simulate_production()
```

```python
def analyze_production():
    print(wells)
    print('CUMULATIVE PRODUCTION FOR WELL(S):')
    for well, well_production in wells_production_rates.items():
        print(f'{well.capitalize()}: {sum(well_production)}')

    print("AVERAGE PRODUCTION(S):")
    for well, well_production in wells_production_rates.items():
        print(f'{well.capitalize()}: {round(sum(well_production) / len(well_production), 2)}')
```

```python
import matplotlib.pyplot as plt
def visualize_data():
    for well_name in wells_production_rates.keys():
        y = wells_production_rates[well_name]
        x = []
        n = well.index(well_name)
        j = years[n]
        for num in range(j + 1):
            x.append(num)

        plt.figure()
        plt.plot(x, y, marker='.')
        plt.xlabel('Time(Year)')
        plt.ylabel('Production Rate')
        plt.title(well_name.upper())
        plt.grid()
        plt.show
```

```python
def main_menu():
    option = input("""
    Choose the number that corresponds with what you would like to do.
    1. Input parameters
    2. Simulate wells
    3. Analyze production
    4. Visualize results
    5. Exit
    """)

    if option.isdigit():
        selected = int(option)
        if selected == 1:
            ask = int(input("How many wells do you want to input parameters for?: "))
            for well in range(ask):
                input_parameters()
            main_menu()
        elif selected == 2:
            simulate_well()
            main_menu()
        elif selected == 3:
            analyze_production()
            main_menu()
        elif selected == 4:
            visualize_data()
            main_menu()
        elif selected == 5:
            print('Exited')
        elif selected > 5:
            print("Input has to be between 1 to 5")
            main_menu()
    else:
        print("Invalid input. Input has to be an integer and should be between 1 to 5")
        main_menu()


main_menu()
```
